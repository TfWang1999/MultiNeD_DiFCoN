# MultiNeD joint calibration

This example calls the same formal entry and paper_settings.py as the command line.
The training cell uses the full 600-epoch paper budget and saves its results in OUTPUT_DIR.


```python
from pathlib import Path
import subprocess
import sys

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "MultiNeD/reproduce_MultiNeD.py").is_file())
DEVICE = "cuda:0"  # Use "cpu" for CPU execution.
OUTPUT_DIR = ROOT / "MultiNeD/outputs/reaction_coupled_run"
DRIFT_SOURCE = None  # Set a new drift run/checkpoint; None uses the supplied weights.
```

Run the next cell to train and save the checkpoint and training history.
The final console line reports the result directory. See docs/REPRODUCIBILITY.md to prepare its figures.


```python
command = [sys.executable, "-B", str(ROOT / "MultiNeD/reproduce_MultiNeD.py"),
           "joint", "--device", DEVICE, "--output-dir", str(OUTPUT_DIR)]
if DRIFT_SOURCE is not None:
    command += ["--drift-source", str(DRIFT_SOURCE)]
subprocess.run(command, cwd=ROOT, check=True)
```
