# Manuscript Figure 2 panels

Purpose: show that the fixed Reaction model reconstructs tumour and CD8 program dynamics, generalizes to strictly held-out tumour timepoints, and retains competitive held-out state-dynamics performance.

Each manuscript panel is generated independently. This notebook reads frozen local plotting tables only. Refresh sources separately with the preparation scripts.


```python
from pathlib import Path
import sys

import matplotlib
matplotlib.use("module://matplotlib_inline.backend_inline")
import matplotlib.pyplot as plt
from IPython.display import display

START_DIR = Path.cwd().resolve()
REPO_DIR = next(
    base for base in (START_DIR, *START_DIR.parents)
    if (base / "MultiNeD/figure_work/scripts/figure_data.py").is_file()
)
sys.path.insert(0, str(REPO_DIR))

from MultiNeD.figure_work.scripts.figure_data import find_analysis_dir, load_main_data, load_main_rollouts
from MultiNeD.figure_work.scripts.figure_style import apply_global_style, get_figure_config, save_figure
from MultiNeD.figure_work.scripts.reaction_figures import (
    plot_loto_predictions, plot_state_reconstruction, plot_tumour_reconstruction,
    plot_full_data_component_ablation_bars, plot_dynamics_benchmark_bars,
    plot_tcell_activity_four_treatments,
)

apply_global_style()
ANALYSIS_DIR = find_analysis_dir()
OUTPUT_DIR = ANALYSIS_DIR / "figures"
print("Analysis directory:", ANALYSIS_DIR)
print("Output directory:", OUTPUT_DIR)
```

## Load local plotting tables

The fixed model evaluation was exported upstream with the original random-state settings.


```python
data = load_main_data(ANALYSIS_DIR)
rollout = load_main_rollouts(ANALYSIS_DIR)
{name: len(frame) for name, frame in data.items()}
```

# Main figure

## Fig. 2A — tumour-volume reconstruction


```python
SAVE_FIGURES = True

cfg = get_figure_config("tumour_reconstruction")
cfg["figsize"] = (100 / 25.4, 60 / 25.4)
cfg["title"] = None
cfg["legend_size"] = 6
cfg["tick_size"] = 8
cfg["label_size"]=8
cfg["title_size"] =8
fig = plot_tumour_reconstruction(rollout["tumour_volume"], cfg, dose_data=rollout["tumour_doses"])
display(fig)
if SAVE_FIGURES:
    save_figure(fig, OUTPUT_DIR, "Fig2A_tumour_reconstruction")
plt.close(fig)
```

## Fig2B T cell dynamic

```python
SAVE_FIGURES = True

fig = plot_tcell_activity_four_treatments(rollout["reaction_volume"])
display(fig)
if SAVE_FIGURES:
    save_figure(fig, OUTPUT_DIR, "Fig2B_baseline_tcell_density_activity")
plt.close(fig)
```

## Fig. 2C — strict LOTO held-out tumour predictions

```python
SAVE_FIGURES = True

cfg = get_figure_config("loto_prediction")
cfg["figsize"] = (105 / 25.4, 58 / 25.4)
cfg["title"] = None
cfg["legend_size"] = 8
cfg["tick_size"] = 8
cfg["xlabel_size"] = 8
cfg["ylabel_size"]=8
fig = plot_loto_predictions(data["loto"], cfg)
display(fig)
if SAVE_FIGURES:
    save_figure(fig, OUTPUT_DIR, "Fig2C_strict_loto")
plt.close(fig)
```

## Fig. 2D — CD8 program-state reconstruction


```python
SAVE_FIGURES = True

cfg = get_figure_config("state_reconstruction")
cfg["figsize"] = (105 / 25.4, 60 / 25.4)
cfg["title"] = None
cfg["legend_size"] = 8
cfg["metric_text_xy"] = (0.48, 0.98)
cfg["tick_size"] = 8
cfg["xlabel_size"] = 8
cfg["ylabel_size"]=8
fig = plot_state_reconstruction(rollout["state_summary"], cfg, z_cols=rollout["z_cols"])
display(fig)
if SAVE_FIGURES:
    save_figure(fig, OUTPUT_DIR, "Fig2D_state_reconstruction")
plt.close(fig)
```

## Fig. 2E — primary held-out dynamics benchmark

```python
SAVE_FIGURES = True

cfg = get_figure_config("dynamics_benchmark_folds")
cfg["figsize"] = (210 / 25.4, 60 / 25.4)
cfg["title"] = None
cfg["tick_size"] = 8
cfg["label_size"] = 8
cfg["title_size"] = 8

fig = plot_dynamics_benchmark_bars(data["dynamics_benchmark_folds"], cfg)
if SAVE_FIGURES:
    save_figure(fig, OUTPUT_DIR, "Fig2E_dynamics_benchmark_barplot",
                bbox_inches=None)
plt.close(fig)
display(fig)
```
