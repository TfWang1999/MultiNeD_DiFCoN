# Policy supplementary figures S6–S7

Run the setup and table-loading cells before the desired panels. Set panel dimensions in millimeters.

```python
from pathlib import Path
import sys

start = Path.cwd().resolve()
root = next(path for path in (start, *start.parents) if (path / "supplementary/supp_plotting").is_dir())
sys.path.insert(0, str(root))

import pandas as pd

from DiFCoN.figure_workbench.plotting._source_tables import read_source_table
from supplementary.supp_plotting.editing import export_panel, figure_size
from supplementary.supp_plotting.paths import BASELINE_DIR, OPENLOOP_DIR, figure_output_dir
from supplementary.supp_plotting.policy import (
    final_score_panel, openloop_mean_test_trajectories_panel, plot_feature_attribution,
    plot_locked_seed_stability, plot_multiseed_metric_overview, plot_training_history_overview,
    plot_validation_openloop_lines,
)

OUTPUT_DIR = figure_output_dir("policy")
```

## Load saved data

```python
baseline_history = pd.read_csv(BASELINE_DIR / "tables/training_history.csv")
final_scores = pd.read_csv(BASELINE_DIR / "tables/test_strategy_summary.csv")
schedules = pd.read_csv(OPENLOOP_DIR / "validation_derived_openloop_schedule.csv")
test_trajectories = read_source_table("openloop", "mean_tumour_trajectories")
seed_outcomes = read_source_table("dose_weight", "full_policy_patient_by_seed")
feature_attribution = read_source_table("attribution", "feature_attribution")
locked_seed_metrics = read_source_table("locked", "seed_metrics").rename(columns={
    "mean_loss": "loss", "mean_final_C_rel": "final_C_rel",
    "mean_tumor_auc_rel": "tumor_auc_rel", "mean_total_dose": "total_dose",
})
```

## S6a–b — Training and validation loss and its components

The left panel shows the training and recorded validation losses. The right panel shows the saved unweighted loss components.

```python
fig = plot_training_history_overview(baseline_history, figsize=figure_size(210, 60), font_size=10)
export_panel(fig, "S6ab_policy_training", output_dir=OUTPUT_DIR)
```

## S6c — Endpoint immune program scores

Means across patients for fixed combination therapy and the baseline policy.

```python
fig = final_score_panel(final_scores, figsize=figure_size(85, 48), font_size=10)
export_panel(fig, "S6c_final_program_scores", output_dir=OUTPUT_DIR)
```

## S6d — Patient metrics across three random seeds

Each seed uses 32 virtual patients. Violin plots show the distributions, and short horizontal lines mark the means.

```python
fig = plot_multiseed_metric_overview(seed_outcomes, figsize=figure_size(125, 48), font_size=10)
export_panel(fig, "S6d_policy_seed_outcomes", output_dir=OUTPUT_DIR)
```

## S6e — Feature and history-block attribution

Show the original integrated-gradient attributions for 16 features on a logarithmic color scale.

```python
fig = plot_feature_attribution(feature_attribution, figsize=figure_size(210, 90), font_size=9.5, cmap="coolwarm")
export_panel(fig, "S6e_input_attribution", output_dir=OUTPUT_DIR)
```

## S7a — Tumor trajectories in the test cohort

Show the saved means across patients and their SEMs.

```python
fig = openloop_mean_test_trajectories_panel(test_trajectories, figsize=figure_size(110, 68), font_size=10)
export_panel(fig, "S7a_openloop_trajectories", output_dir=OUTPUT_DIR)
```

## S7b — Open-loop schedule selected on the validation set

The two drug curves show the mean ± sample SD across the three seeds.

```python
fig = plot_validation_openloop_lines(schedules, figsize=figure_size(100, 68), font_size=10)
export_panel(fig, "S7b_openloop_schedule", output_dir=OUTPUT_DIR)
```

## S7c — Seed stability on the locked test set

Each dot shows the mean for 32 patients at one seed. Bars and error bars show the mean ± sample SD across seeds.

```python
fig = plot_locked_seed_stability(locked_seed_metrics, figsize=figure_size(210, 80), font_size=10)
export_panel(fig, "S7c_benchmark_seed_outcomes", output_dir=OUTPUT_DIR)
```
